# RockCollection

```python
from spacerocks import SpaceRock, RockCollection
from spacerocks.time import Time
from spacerocks.spice import SpiceKernel

import numpy as np

kernel = SpiceKernel()
kernel.load("/Users/thomasruch/.spacerocks/spice/latest_leapseconds.tls")
kernel.load("/Users/thomasruch/.spacerocks/spice/de440s.bsp")
kernel.load("/Users/thomasruch/.spacerocks/spice/earth_1962_240827_2124_combined.bpc")

epoch = Time.now()
```

```python
neptune = SpaceRock.from_spice(name="Neptune Barycenter", epoch=epoch, reference_plane="J2000", origin="SSB")
jupiter = SpaceRock.from_spice(name="Jupiter Barycenter", epoch=epoch, reference_plane="J2000", origin="SSB")
saturn = SpaceRock.from_spice(name="Saturn Barycenter", epoch=epoch, reference_plane="J2000", origin="SSB")
uranus = SpaceRock.from_spice(name="Uranus Barycenter", epoch=epoch, reference_plane="J2000", origin="SSB")
```

```python
planets = RockCollection()
planets.add(neptune)
planets.add(jupiter)
planets.add(saturn)
planets.add(uranus)
```

```python
print(planets)
print(f"a: {planets.a()}")
print(f"e: {planets.e()}")
print(f"inc: {planets.inc()}")
```

```python
planets_beyond_10au = planets.filter(planets.a() > 10)
```

```python
print(planets_beyond_10au)
print(f"a: {planets_beyond_10au.a()}")
print(f"e: {planets_beyond_10au.e()}")
print(f"inc: {planets_beyond_10au.inc()}")
```

```python
from spacerocks.observing import Observatory
```

```python
w84 = Observatory.from_obscode("W84")
observer = w84.at(epoch, reference_plane="J2000", origin="SSB")
o = planets.observe(observer) 
```

```python
print(np.degrees(o[0].ra), np.degrees(o[0].dec))
print(np.degrees(o[1].ra), np.degrees(o[1].dec))
```
