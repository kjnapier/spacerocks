# Transforms

```python
from spacerocks import transforms
```

### The main functionality is to convert to and from conic, mean, and true anomalies

```python
help(transforms.calc_conic_anomaly_from_mean_anomaly)
```

```python
mean_anomaly = 0.5

e_elliptical = 0.5
e_hyperbolic = 1.3
e_parabolic = 1.0
e_circular = 0.0
```

```python
eccentric_anomaly = transforms.calc_conic_anomaly_from_mean_anomaly(e_elliptical, mean_anomaly)
print(f"eccentric_anomaly: {eccentric_anomaly}")

hyperbolic_anomaly = transforms.calc_conic_anomaly_from_mean_anomaly(e_hyperbolic, mean_anomaly)
print(f"hyperbolic_anomaly: {hyperbolic_anomaly}")

parabolic_anomaly = transforms.calc_conic_anomaly_from_mean_anomaly(e_parabolic, mean_anomaly)
print(f"parabolic_anomaly: {parabolic_anomaly}")

circular_anomaly = transforms.calc_conic_anomaly_from_mean_anomaly(e_circular, mean_anomaly)
print(f"circular_anomaly: {circular_anomaly}")
```
